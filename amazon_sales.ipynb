## Part - 2 New column creation

```python
df['order_date'] = pd.to_datetime(df['order_date'],dayfirst=True,format='mixed') 
```

```python
# Extract year from order date
df['year'] = df['order_date'].dt.year
```

```python
# Extract month from order date
df['month'] = df['order_date'].dt.month
```

```python
# Extract month name from order data
df['month_name'] = df['order_date'].dt.month_name()
```

```python
# Extract day from order date
df['day'] = df['order_date'].dt.day
```

```python
df.head()
```

```python
# Create a column called revenue category
def Revenue_category(revenue):
    if revenue>1000:
        return "High Revenue"
    elif revenue>=500:
        return "Medium Revenue"
    else:
        return "Low Revenue"
df['Revenue_category'] = df['total_revenue'].apply(Revenue_category)
df
```

```python
# Create a column called customer satisfaction
def Customer_satisfaction(rating):
    if rating >=4.5:
        return "Excellent"
    elif rating>=3:
        return "Good"
    else:
        return "Poor"
df['Customer_satisfaction'] = df['rating'].apply(Customer_satisfaction)
df
```
